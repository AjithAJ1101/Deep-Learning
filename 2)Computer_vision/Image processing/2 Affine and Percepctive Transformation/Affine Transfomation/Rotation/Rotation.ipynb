```python
import cv2
import numpy
import matplotlib.pyplot as plt
```

```python
#Loading image from memory
image = cv2.imread('chess.png')
```

```python
plt.imshow(image)
```

```python
height, width = image.shape[:2]
```

```python
#Translation matrix
matrix = cv2.getRotationMatrix2D((width/2,height/2), 9, 1)
```

```python
#Applying the matrix to the image 
translated = cv2.warpAffine(image, matrix, (width, height))
```

```python
#Showing the image
cv2.imshow('translation', translated)
cv2.waitKey(4000)
```
