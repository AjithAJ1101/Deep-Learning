# Affine Transformation

1. In Affine transformation, all parallel lines in the original image will still be parallel in the output image. 

2. To find the transformation matrix, we need three points from input image and their corresponding locations in the output image. 



```python

import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
img = cv2.imread('chess.png')
```

```python
rows, cols, ch = img.shape
```

```python
pts1 = np.float32([[50, 50],
                   [200, 50], 
                   [50, 200]])
  
pts2 = np.float32([[10, 100],
                   [200, 50], 
                   [100, 250]])
```

```python
M = cv2.getAffineTransform(pts1, pts2)
dst = cv2.warpAffine(img, M, (cols, rows))
  
```

```python
plt.subplot(121)
plt.imshow(img)
plt.title('Input')
  
plt.subplot(122)
plt.imshow(dst)
plt.title('Output')
  
plt.show()
```
