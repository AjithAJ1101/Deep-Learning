# Affine Transformation

1. In Affine transformation, all parallel lines in the original image will still be parallel in the output image. 

2. To find the transformation matrix, we need three points from input image and their corresponding locations in the output image. 



```python

import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
img = cv2.imread('chess.png')
```

```python
rows, cols, ch = img.shape
```

```python
src_points = np.float32([[0,0], [cols-1,0], [0,rows-1], [cols-1,rows-1]])
dst_points = np.float32([[0,0], [cols-1,0], [int(0.33*cols),rows-1], [int(0.66*cols),rows-1]])
projective_matrix = cv2.getPerspectiveTransform(src_points, dst_points)
img_protran = cv2.warpPerspective(img, projective_matrix, (cols,rows))
```

```python
plt.imshow(img)
```

```python
plt.imshow(img_protran)
```
