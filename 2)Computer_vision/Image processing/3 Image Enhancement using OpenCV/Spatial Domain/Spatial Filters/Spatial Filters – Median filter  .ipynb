# Spatial Filters – Averaging filter and Median filter 

```python
#Low Pass SPatial Domain Filtering
# to observe the blurring effect
 
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the image
img_noisy1  = cv2.imread('2.png', 0)
plt.imshow(img_noisy1 )
```

```python
# Obtain the number of rows and columns
# of the image
m, n = img_noisy1.shape
  
```

```python
# Traverse the image. For every 3X3 area,
# find the median of the pixels and
# replace the center pixel by the median
img_new1 = np.zeros([m, n])
```

```python
for i in range(1, m-1):
    for j in range(1, n-1):
        temp = [img_noisy1[i-1, j-1],
               img_noisy1[i-1, j],
               img_noisy1[i-1, j + 1],
               img_noisy1[i, j-1],
               img_noisy1[i, j],
               img_noisy1[i, j + 1],
               img_noisy1[i + 1, j-1],
               img_noisy1[i + 1, j],
               img_noisy1[i + 1, j + 1]]
         
        temp = sorted(temp)
        img_new1[i, j]= temp[4]
```

```python
img_new1 = img_new1.astype(np.uint8)
```

```python
plt.imshow(img_new1)
```
