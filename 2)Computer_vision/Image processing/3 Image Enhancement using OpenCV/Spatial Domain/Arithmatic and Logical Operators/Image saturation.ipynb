# Image saturation

```python
# Necessary imports
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
# Loading the image
img = cv2.imread('7.jpg',1)
plt.imshow(img)
```

```python
# Creating an image of 1s with the same size 
# as our input image
img_100 = np.ones(img.shape, dtype = "uint8") * 100
# Adding two images
img2=cv2.add(img,img_100)
plt.imshow(img2)
```

```python
# Subtracting two images
img3=cv2.subtract(img,img_100)
plt.imshow(img3)
```
